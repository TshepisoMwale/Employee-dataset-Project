```python
import pandas as pd
```

```python
df= pd.read_excel(r"D:\Cisco Network Academy\Personal\Employee dataset\Original_Employee_dataset.xlsx")
df
```

```python
df.drop_duplicates()
```

```python
df = df.drop(columns = "Status")
df
```

```python
df["Phone"] = df["Phone"].astype("string").str.replace(r'[^0-9]', '', regex=True) # converted the column 'Phone' to a string so that i could remove the unnecessary characters
```

```python
df["Phone"]= df["Phone"].str.replace('[^a-zA-Z0-9]',' ') 
```

```python
df["Phone"] = df["Phone"].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:10] )# added the dashes 
df
```

```python
df[["Department", "Region"]] = df["Department_Region"].str.split('-', expand=True)# split Department_Region into two columns
df
```

```python
df["Age"] = df["Age"].fillna(' ' )
df
```
